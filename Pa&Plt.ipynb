```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Costomers = pd.read_csv('customers.csv')
Product = pd.read_csv('products.csv')
Orders = pd.read_csv('orders.csv')
```

```python
Costomers.head()
```

```python
Costomers.head().shape
```

```python
Product.head()
```

```python
Orders.head()
```

```python
order_product=pd.merge(Orders,Product,on='ProductID')
order_product
```

```python
df_master=pd.merge(order_product,Costomers,on='CustomerID')
df_master.head()
```

```python
df_master['UnitPrice']=df_master['Sales'] / df_master['Quantity']
df_master.head()
```

```python
df_master['Margin']=df_master['Profit'] / df_master['Sales']
df_master.head()
```

```python
df_master['OrderDate']=pd.to_datetime(df_master['OrderDate'])
df_master.info()
```

```python
df_master2=df_master.set_index('OrderDate')
df_master2.head()
```

```python
df_master2[['Sales','Profit']].resample('ME').sum()
```

```python
x=df_master2['Sales']
y=df_master2['Profit']
plt.plot(x,color='blue',label='sale',alpha=0.4)
plt.plot(y,color='green',label='profit',alpha=0.4)
plt.title('Monthly Sales vs Profit')
plt.xlabel('sale')
plt.ylabel('profit')
plt.legend()
plt.show()
```

```python
sns.boxplot(x='Category',y='Profit',data=df_master2)
plt.title('boxplot profit by category')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.show()
```
